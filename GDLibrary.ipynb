```python
#BATCH GRADIENT DESCENT
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

X,y=make_regression(n_samples=100,
                    n_features=2,
                    n_informative=2,
                    n_targets=1,
                    noise=50)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

lr=LinearRegression()
lr.fit(X_train,y_train)
predict1=lr.predict(X_test)
print(f"LR_Intercept: {lr.intercept_}, LR_Coefficient: {lr.coef_}")
print(f"LinearRegression_r2_score: {r2_score(y_test, predict1)}")

class BatchGD:
    def __init__(self,learning_rate,epochs):
        self.intercept_=None
        self.coef_=None
        self.lr=learning_rate
        self.epochs=epochs

    def fit(self,X_train,y_train):
        self.intercept_=0
        self.coef_=np.ones(X_train.shape[1])

        for i in range(self.epochs):
            #Calculation for intercept
            y_pred=self.intercept_ + np.dot(X_train,self.coef_)
            intercept_der=-2*np.mean(y_train-y_pred)
            self.intercept_=self.intercept_-self.lr*intercept_der

            #Calculation for coefficients
            coef_der= -2*(np.dot(X_train.T,y_train-y_pred)/X_train.shape[0])
            self.coef_=self.coef_ - self.lr*coef_der

        print(f"BGD_Intercept: {lr.intercept_}, BGD_Coefficient: {lr.coef_}")

    def predict(self,X_test):
        return np.dot(X_test,self.coef_)+self.intercept_

bgd=BatchGD(learning_rate=0.5,epochs=100)
bgd.fit(X_train,y_train)
predict2=bgd.predict(X_test)
print(f"BGD_r2_score: {r2_score(y_test,predict2)}")
```

```python
#STOCHASTIC GRADIENT DESCENT
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

X,y=load_diabetes(return_X_y=True)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
lr=LinearRegression()
lr.fit(X_train,y_train)
predict1=lr.predict(X_test)
print(f"LR_Intercept: {lr.intercept_}, LR_Coefficient: {lr.coef_}")
print(f"LinearRegression_r2_score: {r2_score(y_test, predict1)}")

class S_Gradient:
    def __init__(self,learning_rate,epochs):
        self.intercept_=None
        self.coef_=None
        self.lr=learning_rate
        self.epochs=epochs

    def fit(self,X_train,y_train):
        self.intercept_=0
        self.coef_=np.ones(X_train.shape[1])

        for i in range(self.epochs):
            for j in range(X_train.shape[0]):
                idx=np.random.randint(0,X_train.shape[0])

                y_pred=self.intercept_+np.dot(X_train[idx],self.coef_)
                intercept_der= -2 * (y_train[idx]-y_pred)
                self.intercept_=self.intercept_-self.lr*intercept_der

                coef_der=-2*np.dot(y_train[idx]-y_pred,X_train[idx])
                self.coef_=self.coef_-self.lr*coef_der

        print(f"SGD_Intercept: {self.intercept_}, SGD_Coefficient: {self.coef_}")

    def predict(self,X_test):
        return np.dot(X_test,self.coef_)+self.intercept_

sgd=S_Gradient(learning_rate=0.01,epochs=150)
sgd.fit(X_train,y_train)
predict2=sgd.predict(X_test)
print(f"SGD_r2_score: {r2_score(y_test,predict2)}")
```

```python
#MINI BATCH GRADIENT DESCENT
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

X,y=load_diabetes(return_X_y=True)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

lr=LinearRegression()
lr.fit(X_train,y_train)
predict1=lr.predict(X_test)
print(f"LR_Intercept: {lr.intercept_}, LR_Coefficient: {lr.coef_}")
print(f"LinearRegression_r2_score: {r2_score(y_test, predict1)}")

class mini_Batch:
    def __init__(self,learning_rate,epochs,batch_size):
        self.coef_=None
        self.intercept_=None
        self.lr=learning_rate
        self.epochs=epochs
        self.batch_size=batch_size

    def fit(self,X_train,y_train):
        self.coef_=np.ones(X_train.shape[1])
        self.intercept_=0

        for i in range(self.epochs):
            for j in range(X_train.shape[0]//self.batch_size):
                idx=np.random.choice(X_train.shape[0],size=self.batch_size,replace=False)

                y_pred=self.intercept_+np.dot(X_train[idx],self.coef_)
                intercept_der=-2*np.mean(y_train[idx]-y_pred)
                self.intercept_=self.intercept_-self.lr*intercept_der

                coef_der=-2*np.dot(X_train[idx].T,y_train[idx]-y_pred)/self.batch_size
                self.coef_=self.coef_-self.lr*coef_der
        print(f"mini_Batch Intercept: {self.intercept_}, mini_Batch coefficient: {self.coef_}")

    def predict(self,X_test):
        return self.intercept_+np.dot(X_test,self.coef_)

mb=mini_Batch(learning_rate=0.07,epochs=200,batch_size=X_train.shape[0]//10)
mb.fit(X_train,y_train)
predict2=mb.predict(X_test)
print(f"mini_Batch_r2_score: {r2_score(y_test,predict2)}")
```
