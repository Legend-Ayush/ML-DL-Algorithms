# Toy Neural Network

### Structure and Procedure
- Input layer: 2 neurons
- Hidden layer: 2 neurons
- Ouput layer activation: Sigmoid 
- Output layer: 1 neuron
- Hidden layer activation: Linear 
- Loss: \(1/2(y-\hat y)^2\)
- Backpropagation: chain rule
- Optimizer: Gradient Descent
- Parameters: \(W_1,b_1,W_2,b_2\)

-NumPy: only

```python
import numpy as np

x=np.array([2.0, 3.0])
y=1

W1=np.array([
    [0.5, 0.4],
    [0.3, 0.2]
])
b1=np.array([0.1, 0.2])

W2=np.array([0.6, 0.4])
b2=0.1

lr=0.1
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
z1=W1.T@x + b1
a1=sigmoid(z1)

y_hat=W2@a1 + b2

loss=0.5*(y-y_hat)**2

print("z1 =", z1)
print("a1 =", a1)
print("y_hat =", y_hat)
print("loss =", loss)
```

```python
# Output layer gradients
d_y_hat = y_hat - y
dW2 = d_y_hat * a1
db2 = d_y_hat

# Hidden layer gradients
da1 = d_y_hat * W2
dz1 = da1 * a1 * (1 - a1)
dW1 = np.outer(x, dz1)
db1 = dz1
```

```python
W2 = W2 - lr * dW2
b2 = b2 - lr * db2

W1 = W1 - lr * dW1
b1 = b1 - lr * db1


print("Updated W1:\n", W1)
print("Updated b1:", b1)

print("Updated W2:", W2)
print("Updated b2:", b2)
```
